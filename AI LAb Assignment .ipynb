```python
students = []

def result(a, b, c):
    total = a + b + c
    per = total / 3

    if per >= 80:
        grade = "A"
    elif per >= 70:
        grade = "B"
    elif per >= 60:
        grade = "C"
    elif per >= 50:
        grade = "D"
    else:
        grade = "F"

    return total, per, grade


for i in range(5):
    name = input("Name: ")
    roll = input("Roll No: ")
    p = int(input("Python marks: "))
    ai = int(input("AI marks: "))
    m = int(input("Math marks: "))

    total, per, grade = result(p, ai, m)

    students.append({
        "name": name, "roll": roll,
        "total": total, "per": per, "grade": grade
    })


for s in students:
    print(s)


top = max(students, key=lambda x: x["per"])
print("Highest:", top["name"], top["per"])


with open("results.txt", "w") as f:
    for s in students:
        f.write(str(s) + "\n")
        
```

```python
def analysis(text):
    words = text.split()

    print("Words:", len(words))
    print("Characters:", len(text))
    print("Longest:", max(words, key=len))
    print("Shortest:", min(words, key=len))

    vowels = 0
    for ch in text.lower():
        if ch in "aeiou":
            vowels += 1

    print("Vowels:", vowels)

    unique = set(words)
    print("Unique words:", unique)
    print("Alphabetical:", sorted(unique))

    with open("word_analysis.txt", "w") as f:
        f.write("Words: " + str(len(words)) + "\n")
        f.write("Characters: " + str(len(text)) + "\n")
        f.write("Longest: " + max(words, key=len) + "\n")
        f.write("Shortest: " + min(words, key=len) + "\n")
        f.write("Vowels: " + str(vowels) + "\n")
        f.write("Words: " + str(sorted(unique)))


text = input("Enter sentence: ")
analysis(text)
```

```python
products = []

def bill(price, quantity):
    return price * quantity

for i in range(5):
    name = input("Product name: ")
    price = float(input("Price: "))
    quantity = int(input("Quantity: "))

    total = bill(price, quantity)

    products.append({
        "name": name,
        "price": price,
        "quantity": quantity,
        "total": total
    })

total_bill = sum(p["total"] for p in products)

if total_bill > 10000:
    discount = total_bill * 0.10
else:
    discount = 0

final = total_bill - discount

expensive = max(products, key=lambda x: x["price"])

print("Bill:", total_bill)
print("Discount:", discount)
print("Final Bill:", final)
print("Expensive:", expensive["name"])

with open("bill.txt", "w") as f:
    f.write("Total Bill: " + str(total_bill) + "\n")
    f.write("Discount: " + str(discount) + "\n")
    f.write("Final Bill: " + str(final) + "\n")
    f.write("Expensive Product: " + expensive["name"])
```

```python
def add(a, b):
    return a + b

def sub(a, b):
    return a - b

def mul(a, b):
    return a * b

def div(a, b):
    return a / b

def mod(a, b):
    return a % b


while True:
    a = float(input("First number: "))
    b = float(input("Second number: "))

    op = input("Enter +, -, *, / or %: ")

    if op == "+":
        print("Result:", add(a, b))
    elif op == "-":
        print("Result:", sub(a, b))
    elif op == "*":
        print("Result:", mul(a, b))
    elif op == "/":
        print("Result:", div(a, b))
    elif op == "%":
        print("Result:", mod(a, b))
    else:
        print("Wrong operation")

    again = input("Again? yes/no: ")

    if again.lower() != "yes":
        break
```

```python
menu = {
    1: ["Burger", 500],
    2: ["Pizza", 800],
    3: ["Fries", 250],
    4: ["Drink", 150]
}

def price(p, q):
    return p * q

orders = []
total = 0

while True:
    print("\n1. Burger - 500")
    print("2. Pizza - 800")
    print("3. Fries - 250")
    print("4. Drink - 150")
    print("5. Finish")

    choice = int(input("Choose: "))

    if choice == 5:
        break

    if choice in menu:
        quantity = int(input("Quantity: "))

        name = menu[choice][0]
        p = menu[choice][1]

        amount = price(p, quantity)
        total += amount

        orders.append(name + " x " + str(quantity))

    else:
        print("Wrong choice")

print("\nOrders:")

for order in orders:
    print(order)

print("Final Bill:", total)

with open("orders.txt", "w") as f:
    for order in orders:
        f.write(order + "\n")

    f.write("Final Bill: " + str(total))
```
